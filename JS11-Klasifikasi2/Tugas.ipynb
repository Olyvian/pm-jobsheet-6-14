```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import accuracy_score

# 1. SVM pada voice.csv dengan berbagai kernel dan split
data = pd.read_csv('voice.csv')
X = data.drop('label', axis=1)
y = LabelEncoder().fit_transform(data['label'])

for split in [0.3, 0.2]:
    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=split, random_state=42)
    sc = StandardScaler()
    Xtr = sc.fit_transform(Xtr); Xte = sc.transform(Xte)
    for kernel in ['linear', 'poly', 'rbf']:
        m = SVC(kernel=kernel)
        m.fit(Xtr, ytr)
        print(f'split test={split} kernel={kernel} acc={accuracy_score(yte, m.predict(Xte)):.4f}')

```
