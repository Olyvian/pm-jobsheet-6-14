# JS09 - Klasifikasi: Tugas 2 (spam: CountVectorizer vs TF-IDF, stop_words)



```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

df = pd.read_csv('spam.csv', encoding='latin-1')
df = df.drop(df.iloc[:, 2:], axis=1)
df['v1'] = df['v1'].map({'spam': 1, 'ham': 0})
X = df['v2'].values
y = df['v1'].values
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# CountVectorizer dengan stop_words
cv = CountVectorizer(stop_words='english')
Xtr = cv.fit_transform(X_train); Xte = cv.transform(X_test)
m = MultinomialNB(); m.fit(Xtr, y_train)
print('CountVectorizer stop_words - train:', accuracy_score(y_train, m.predict(Xtr)), 'test:', accuracy_score(y_test, m.predict(Xte)))

# TF-IDF dengan stop_words
tfidf = TfidfVectorizer(stop_words='english')
Xtr2 = tfidf.fit_transform(X_train); Xte2 = tfidf.transform(X_test)
m2 = MultinomialNB(); m2.fit(Xtr2, y_train)
print('TF-IDF stop_words - train:', accuracy_score(y_train, m2.predict(Xtr2)), 'test:', accuracy_score(y_test, m2.predict(Xte2)))
```
