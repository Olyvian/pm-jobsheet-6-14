# JS09 - Klasifikasi: Tugas 1 (kNN pada voice.csv)



```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

data = pd.read_csv('voice.csv')
X = data.drop('label', axis=1)
y = LabelEncoder().fit_transform(data['label'])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

# coba beberapa nilai K
accs = []
for k in range(1, 21):
    m = KNeighborsClassifier(n_neighbors=k)
    m.fit(X_train_s, y_train)
    accs.append(m.score(X_test_s, y_test))

best_k = int(np.argmax(accs) + 1)
print('Best K:', best_k, 'Acc:', max(accs))

import matplotlib.pyplot as plt
plt.plot(range(1, 21), accs, marker='o')
plt.xlabel('K'); plt.ylabel('Akurasi'); plt.title('K vs Akurasi (voice)')
plt.show()
```
